Импортируем инструменты:
pandas — для профессиональной работы с табличными данными
re — для поиска и удаления текстового мусора по сложным шаблонам
os — для управления путями и файлами в системе


```python
import pandas as pd
import re
import os

```

=== БЛОК 1: НАСТРОЙКА ПУТЕЙ ===
Указываем программе, где взять очищенные данные и куда положить финальный результат для заказчика


```python
input_path = r"C:\Users\user\Desktop\Работа\Почти финал\literaguru_strict_clean.csv"
output_path = r"C:\Users\user\Desktop\Работа\Почти финал\literaguru_final_for_customer.csv"

```

=== БЛОК 2: ЗАГРУЗКА ДАННЫХ ===
Читаем файл. Используем utf-8-sig, чтобы русский язык корректно отображался во всех редакторах


```python
df = pd.read_csv(input_path, sep=';', encoding='utf-8-sig')


# === БЛОК 3: ФУНКЦИЯ ФИНАЛЬНОЙ "ПОЛИРОВКИ" ТЕКСТА ===
def final_clean(text):
    # Если в ячейке нет текста, возвращаем её как есть
    if not isinstance(text, str):
        return text

```

1. Удаляем "технические хвосты", которые обычно идут в конце статьи
Мы ищем ключевые слова (Источник, Ссылки, Рубрика и т.д.) и отрезаем всё, что идет после них


```python
    patterns_to_cut = [
        r'Источник\s*:.*',      # Удаление упоминания источника
        r'https?://\S+',        # Удаление любых интернет-ссылок
        r'Рубрика\s*:.*',       # Удаление категории сайта
        r'Метки\s*:.*',         # Удаление тегов
        r'Читать далее.*',      # Удаление кнопок перехода
        r'Подробнее.*'          # Удаление призывов к чтению
    ]

    for pattern in patterns_to_cut:
        # Разбиваем текст по шаблону и оставляем только первую часть (до мусора)
        text = re.split(pattern, text, flags=re.IGNORECASE)[0]

```

2. Удаляем упоминания названия сайта и авторских прав (брендирование)
Это нужно, чтобы текст выглядел нейтральным и чистым


```python
    text = re.sub(r'©\s*Литерагуру', '', text, flags=re.IGNORECASE)
    text = re.sub(r'Литерагуру', '', text, flags=re.IGNORECASE)

    # 3. Финальная стрижка: убираем лишние пробелы и пустые строки по краям
    return text.strip()


```

=== БЛОК 4: ЗАПУСК ОЧИСТКИ ===
Применяем созданную функцию только к основной колонке с текстом сочинения


```python
df['Текст'] = df['Текст'].apply(final_clean)

```

=== БЛОК 5: СОХРАНЕНИЕ И ОТЧЕТ ===
Сохраняем финальный результат в CSV.
Используем кодировку utf-8-sig, чтобы Excel у заказчика открыл файл сразу, без иероглифов и настройки импорта


```python
df.to_csv(output_path, sep=';', index=False, encoding='utf-8-sig')

# Вывод краткой статистики для контроля выполнения
print(f"Очистка завершена. Файл сохранен: {output_path}")
print(f"Обработано строк: {len(df)}")
```
